```python
import mlflow
import mlflow.sklearn
import joblib
import pandas as pd
import mlflow.lightgbm

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
```

```python
X_train = pd.read_csv("../data/X_train.csv")
X_test = pd.read_csv("../data/X_test.csv")

y_train = pd.read_csv("../data/y_train.csv").squeeze()
y_test = pd.read_csv("../data/y_test.csv").squeeze()
```

```python
model = joblib.load("../models/lightgbm_tuned.pkl")
```

```python
import mlflow

mlflow.set_tracking_uri("sqlite:///../mlflow.db")
```

```python
mlflow.set_experiment("Student Dropout Prediction")
with mlflow.start_run():

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average="macro")
    recall = recall_score(y_test, y_pred, average="macro")
    f1 = f1_score(y_test, y_pred, average="macro")

    # Log parameters
    mlflow.log_param("Model", "LightGBM")

    # Log metrics
    mlflow.log_metric("Accuracy", accuracy)
    mlflow.log_metric("Precision", precision)
    mlflow.log_metric("Recall", recall)
    mlflow.log_metric("F1 Score", f1)

    # Save model
    mlflow.lightgbm.log_model(
        model,
        artifact_path="model"
    )

print("Experiment Logged Successfully!")
```

```python
import mlflow
import joblib
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Tracking URI
mlflow.set_tracking_uri("sqlite:///../mlflow.db")

# Experiment
mlflow.set_experiment("Student Dropout Prediction")

# Load data
X_train = pd.read_csv("../data/X_train.csv")
X_test = pd.read_csv("../data/X_test.csv")

y_train = pd.read_csv("../data/y_train.csv").squeeze()
y_test = pd.read_csv("../data/y_test.csv").squeeze()

# Load tuned Random Forest
model = joblib.load("../models/random_forest_tuned.pkl")

with mlflow.start_run(run_name="Random Forest"):

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average="macro")
    recall = recall_score(y_test, y_pred, average="macro")
    f1 = f1_score(y_test, y_pred, average="macro")

    # Parameters
    mlflow.log_param("Model", "Random Forest")

    # Metrics
    mlflow.log_metric("Accuracy", accuracy)
    mlflow.log_metric("Precision", precision)
    mlflow.log_metric("Recall", recall)
    mlflow.log_metric("F1 Score", f1)

    # Save model
    joblib.dump(model, "../models/random_forest_tuned.pkl")

    # Log artifact
    mlflow.log_artifact("../models/random_forest_tuned.pkl")

print("Random Forest Experiment Logged!")
```
